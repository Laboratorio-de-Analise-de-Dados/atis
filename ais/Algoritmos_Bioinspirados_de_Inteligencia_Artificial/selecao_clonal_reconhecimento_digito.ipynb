#### Função anticorpo

Quero construir uma espaço 11 X 11 então num range de 121 vou gerar valores aleatórios entre 0 e 1

```python
def anticorpo() -> list:
    posicoes = []
    for i in range(121):
        posicoes.append(random.randint(0,1))
    return posicoes
```

### População de anticorpos

```python
def populacao(quant: int) -> list:
    """
    Gera populações-solução de anticorpos.
    quant: número de populações-solução a serem criadas
    """
    populacao = []
    for i in range(quant):
        populacao.append(anticorpo())
    return populacao
```

```python
pop = populacao(5)

for i in pop:
    print(i)
```

### Função de aptidão
Ela vai avaliar se os candidatos são bons ou não 

```python
def afinidade(anticorpo: list, antigeno: list) -> int:
    score = 0
    count = 0
    while count < len(anticorpo): # percorrer todo o vetor anticorpo
        if anticorpo[count] == antigeno[count]: # verificar se o index do anticorpo é igual o index do antígeno
            score += 1 
        count += 1
    return score
```

```python
a = anticorpo()

print(f"anticorpo:\n{a}")
print(f"\nnumero a ser reconhecido:\n{numero}")
```

```python
print(afinidade(a,numero))
```

### Lista de afinidade Anticorpo-Antígeno

```python
def afin_anticorpos(pop_anticorpos: list):
    afin_anticorpos = []
    for i in pop_anticorpos:
        afin_anticorpo = []
        afin_anticorpo.append(afinidade(i, numero))
        afin_anticorpo.append(i)
        afin_anticorpos.append(afin_anticorpo)
    return afin_anticorpos
```

```python
pop = populacao(20)
afin = afin_anticorpos(pop)
for x in afin:
    print(x)
```

### Seleção dos N melhores anticorpos

```python
def selec(afinidade: int, numeroAB: int): # numeroAB: Número de anticorpos a serem selecionados
    ordenado = sorted(afinidade) # ordenar por ordem de afinidade
    reverso = ordenado[::-1] # ordenar por ordem decrescente 
    melhores = reverso[:numeroAB] # selecionar os com afinidade mais alta de acordo com o número escolhido
    return melhores
```

```python
pop = populacao(20)
afin = afin_anticorpos(pop)
sel = selec(afin,3)
for i in sel:
    print(i)

print()

for i in afin:
    print(i)
```

## Implementação de algoritmos de Clonagem e Hipermutação
---

### Clonagem
$$QC^{i}_{k} = (\frac{af^{i}_{k}}{\sum^{n}_{k=1}af{i}_{k}})*Cl$$

```python
def clonagem(melhores, clones):
    n_clones = []
    afin = [x[0] for x in melhores] # primeiro termo da lista (afinidade)
    som_afin = sum(afin)
    anti = [x[1] for x in melhores] # segundo termo da lista (anticorpo)
    clonagem = [round((c/som_afin)*clones) for c in afin]

    count = 0
    while count < len(clonagem):
        for i in range(clonagem[count]):
            dobro = []
            dobro.append(afin[count])
            dobro.append(anti[count])
            n_clones.append(dobro)
        count += 1
    return n_clones
```

```python
pop = populacao(20)
afin = afin_anticorpos(pop)
sel = selec(afin,3)
clonado = clonagem(sel, 17)

print("Anticorpos clonados:\n")
for i in clonado:
    print(i[1])

print()

print("Todos os Anticorpos com suas afinidades:\n")
for i in afin:
    print(f"{i}")

print(len(clonado))
```

### Hipermutação
$$hipAnt^{i}_{k} = (1-(\frac{afin^{i}_{k}}{121}))*\beta$$

```python
def hipermutacao(clones, fator_mutacao):
    """
    fator_mutacao = [0.1; 1; 10...]
    """
    hipermutados = []
    clones = [x[1] for x in clones]
    for i in clones:
        aff = afinidade(i, numero)
        taxa_mutacao = (1-(aff/121)) * fator_mutacao
        hipermut = []
        for j in i:
            if random.random() <= taxa_mutacao:
                hipermut.append(1-j)
            else:
                hipermut.append(j)
        hipermutados.append(hipermut)
    return hipermutados
            
```

```python
pop = populacao(20)
afin = afin_anticorpos(pop)
sel = selec(afin,3)
clonado = clonagem(sel, 17)

for i in clonado:
    print(i)

print()

for i in hipermutacao(clones=clonado, fator_mutacao=0.1):
    print(i)
```

```python
pop = populacao(30)
print("pop inicial:", len(pop[0]))

for i in range(100):
    afinidades = afin_anticorpos(pop_anticorpos=pop)
    melhores = selec(afinidade=afinidades, numeroAB=4)
    print("melhores:", melhores[0])
    print(melhores[0])

    clon = clonagem(melhores=melhores, clones=25)
    print("clones:", len(clon[0]))

    hipermut = hipermutacao(clones=clon, fator_mutacao=0.1)
    print("hipermut:", len(hipermut[0]))

    nNovos = len(pop) - len(hipermut)
    for i in range(nNovos):
        hipermut.append(anticorpo())
    pop = hipermut
    print(pop[0])
    print(afinidade(pop[0], numero))
    print()

    imagem = pop[0]
    imagem_matriz = []
    inicio = 0
    fim = 11
    print(type(imagem), len(imagem))
    while fim < (len(imagem)-1):
        matriz = imagem[inicio:fim]
        imagem_matriz.append(matriz)
        inicio += 11
        fim += 11

    imagem = np.array(imagem_matriz)
    fig,ax = plt.subplots()
    im = ax.imshow(imagem)
    ax.set_title('Número')
    fig.tight_layout()
    plt.show()
```

```python
pop = populacao(30)
historico = []  # (geracao, anticorpo, afinidade)

for geracao in range(100):
    afinidades = afin_anticorpos(pop_anticorpos=pop)
    melhores = selec(afinidade=afinidades, numeroAB=4)
    clon = clonagem(melhores=melhores, clones=25)
    hipermut = hipermutacao(clones=clon, fator_mutacao=0.1)
    nNovos = len(pop) - len(hipermut)
    for _ in range(nNovos):
        hipermut.append(anticorpo())
    pop = hipermut

    melhor = pop[0]
    historico.append((geracao, list(melhor), afinidade(melhor, numero)))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

alvo = np.array(numero).reshape(11, 11)

fig, (ax_alvo, ax_ab) = plt.subplots(1, 2, figsize=(7, 3.8))

ax_alvo.imshow(alvo, cmap="gray_r", vmin=0, vmax=1)
ax_alvo.set_title("Antígeno")
ax_alvo.axis("off")

im = ax_ab.imshow(np.zeros((11, 11)), cmap="gray_r", vmin=0, vmax=1)
titulo = ax_ab.set_title("")
ax_ab.axis("off")

def atualizar(frame):
    geracao, ab, afin = historico[frame]
    im.set_data(np.array(ab).reshape(11, 11))
    titulo.set_text(f"Geração {geracao} | Afinidade {afin}/121")
    return im, titulo

anim = FuncAnimation(fig, atualizar, frames=len(historico), interval=150, blit=False)
plt.close(fig)               
HTML(anim.to_jshtml())       
```
